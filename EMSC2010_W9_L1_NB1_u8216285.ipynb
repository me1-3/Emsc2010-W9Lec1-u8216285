# EMSC2010-W9-L1-NB1
Model Earth's surface temperature (in Kelvin) using a zero-dimensional energy balance model without an atmosphere.

$T_e = \left[ \frac{S}{4 \alpha} (1-A)\right]^{1/4}$

where:
* $T_e$ is Earth's surface temperature in Kelvin
* $S$ is the solar flux (1366 W/m$^2$)
* $A$ is albedo (currently around 0.3)
* $\alpha$ is the Stefan-Boltmann constant (5.67E-8 W/m$^2$/K$^4$)

**Modelling greenhouse effect - what if none & how strong is it?**

```python
S = 1366
A = 0.3
alpha = 5.67E-8
```

```python
T_e = ((S/(4*alpha))*(1-A))**(1/4)
```

```python
print(T_e)
```

```python
S = 1366
A = 0 #what if temp difference albedo goes to 0?
alpha = 5.67E-8
T_e2 = ((S/(4*alpha))*(1-A))**(1/4)
print(T_e2)
diff = T_e2 - T_e
print(diff)
```

```python
S = 1366*0.9 #what if solar flux dimmed by 10%?
A = 0.3
alpha = 5.67E-8
T_e3 = ((S/(4*alpha))*(1-A))**(1/4)
print(T_e3)
diff = T_e3 - T_e
print(diff)
```
