#### The logistic population model with harvesting:

$x_{\mathrm{next}} = a x_{\mathrm{initial}}(1-x_{\mathrm{initial}}) - h$

where:
* $x$ is the population size between 0 and 1.
* $x_{\mathrm{initial}}$ is the current population size.
* $x_{\mathrm{next}}$ is next year's population size.
* $a$ is the ability to reproduce.
* $h$ is the size of the harvest between 0 and 1.


```python
### Code the logistic population model with harvesting
#SETUP --- chaotic system -- bc a tiny change creates dramaticaly diff outcomes
a = 3.4
h = 0.2 #take some out of the system
x = np.zeros(50)
x[0] = 0.9
#FORLOOP
for i in range(1,50):
  x[i] = a*x[i-1]*(1-x[i-1]) - h #harvest
#PLOT
plt.plot(x)
plt.xlabel('yrs')
plt.ylabel('pop size 0-1') #where 0 is extinct and 1 is max pop
plt.title('pop dynamics')
```

```python
### Code the logistic population model with harvesting
#SETUP --- chaotic system -- bc a tiny change creates dramaticaly diff outcomes
a = 2.4
h = 0.2 #take some out of the system
x = np.zeros(50)
x[0] = 0.5
#FORLOOP
for i in range(1,50):
  x[i] = a*x[i-1]*(1-x[i-1]) - h #harvest
  x[i] = np.maximum(x[i], 0) #whatever one is bigger is shown, so if x[i] negative, then num = 0

#PLOT
plt.plot(x)
plt.xlabel('yrs')
plt.ylabel('pop size 0-1') #where 0 is extinct and 1 is max pop
plt.title('pop dynamics')
plt.xlim(0,50)
plt.ylim(0,1)
```

#### The logistic population model with randomness:

$x_{\mathrm{next}} = (a+r) x_{\mathrm{initial}}(1-x_{\mathrm{initial}})$

where:
* $x$ is the population size between 0 and 1.
* $x_{\mathrm{initial}}$ is the current population size.
* $x_{\mathrm{next}}$ is next year's population size.
* $a$ is the ability to reproduce.
* $r$ is a normally distributed random number that changes each year.

```python
### Code the logistic population model with randomness

#SETUP --- chaotic system -- bc a tiny change creates dramaticaly diff outcomes
a = 2.4
h = 0.0 #take some out of the system
x = np.zeros(50)
x[0] = 0.5
#FORLOOP
for i in range(1,50):
  r = np.random.randn()*0.3 #times by standard deviations
  x[i] = (a+r)*x[i-1]*(1-x[i-1]) - h #harvest
  x[i] = np.maximum(x[i], 0) #whatever one is bigger is shown, so if x[i] negative, then num = 0

#PLOT
plt.plot(x)
plt.xlabel('yrs')
plt.ylabel('pop size 0-1') #where 0 is extinct and 1 is max pop
plt.title('pop dynamics')
plt.xlim(0,50)
plt.ylim(0,1)
```
